```python
import pandas as pd

df = pd.read_csv("Telco_Customer_Churn_Dataset  (3).csv")
```

```python
df.head(10)    #Checkimng the first 10 rows of the dataset
```

```python
df.dtypes       #Checking the data types of each column in the dataset
```

```python
df.shape      # counting the number of rows and columns in the dataset
```

```python
#changing the data type of the column "TotalCharges" to float64

df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")  
print(df["TotalCharges"].dtype)          # should now say float64
print(df["TotalCharges"].isnull().sum()) # should show 11
```

```python
df.isnull().sum()    #Checking for missing values in the dataset
```

```python
df["TotalCharges"] = df["TotalCharges"].fillna(0)  # Filling missing values with 0
```

```python
print(df["TotalCharges"].dtype)   # must say float64
```

```python
#Checking the duplicates
print("Duplicate rows:", df.duplicated().sum())
```

```python
# checking the column names and changing them to lower case and replacing spaces with underscores

df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")    
df.columns 
```

```python
print(df.isnull().sum().sum())   # should be 0
df.info()
```

```python
# Saving the cleaned dataset to a new CSV file named "telco_cleaned.csv" without the index column.
df.to_csv("telco_cleaned.csv", index=False)    
```
